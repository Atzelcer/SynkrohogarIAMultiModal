## Uso general
El modelo procesa texto en español y devuelve un vector WAV a 22050 Hz.

```python
import librosa
import soundfile as sf
import numpy as np
from scipy.signal import butter, sosfilt, wiener

def enhance_audio(input_path, output_path, target_sr=22050):

    # Cargar audio generado por Tacotron2
    audio, sr = librosa.load(input_path, sr=None)

    # Resample a mayor calidad
    audio = librosa.resample(audio, orig_sr=sr, target_sr=target_sr)

    # Normalizar volumen
    audio = audio / np.max(np.abs(audio)) * 0.98

    # High-boost simple usando filtro high-pass
    # Frecuencia de corte: 3500 Hz
    cutoff = 3500
    sos = butter(2, cutoff, btype="highpass", fs=target_sr, output="sos")
    high = sosfilt(sos, audio)

    # Mezcla del realce de altas frecuencias
    audio = audio + 0.25 * high

    # Reducción de ruido leve
    audio = wiener(audio)

    # Guardar audio nítido
    sf.write(output_path, audio, target_sr)
```

```python
import torch
from TTS.api import TTS
import numpy as np
import sounddevice as sd
import librosa

# ---------------------------------------------------
# CONFIGURACIÓN
# ---------------------------------------------------
CONFIG = {
    "speed": 1.0,
    "pitch": 1.0,
    "volume": 1.15,
    "normalize": True,
    "sample_rate_out": 48000
}

# ---------------------------------------------------
# CARGAR MODELO
# ---------------------------------------------------
device = "cuda" if torch.cuda.is_available() else "cpu"
tts = TTS("tts_models/es/css10/vits").to(device)

# ---------------------------------------------------
# NORMALIZACIÓN
# ---------------------------------------------------
def normalize_audio(audio):
    max_val = np.max(np.abs(audio))
    if max_val > 0:
        audio = audio / max_val
    return audio

# ---------------------------------------------------
# FUNCIÓN FINAL DE AUDIO
# ---------------------------------------------------
def reproducir_texto(texto):
    print("\nTexto recibido:", texto)

    # 1. TTS
    audio = tts.tts(texto, speed=CONFIG["speed"], pitch=CONFIG["pitch"])
    audio = np.array(audio, dtype=np.float32)

    # 2. Volumen
    audio = audio * CONFIG["volume"]

    # 3. Normalizar
    if CONFIG["normalize"]:
        audio = normalize_audio(audio)

    # 4. Resample 48kHz
    sr = CONFIG["sample_rate_out"]
    audio = librosa.resample(audio, orig_sr=22050, target_sr=sr)

    # 5. Cola extra para evitar cortes
    extra_time = 0.40   # 400ms
    extra_samples = int(extra_time * sr)
    cola = np.zeros(extra_samples, dtype=np.float32)

    audio = np.concatenate([audio, cola])

    # 6. Reproducir
    sd.play(audio, sr)
    sd.wait()

    print("Audio finalizado.\n")

# ---------------------------------------------------
# LOOP INTERACTIVO
# ---------------------------------------------------
while True:
    texto = input("Escribe un texto para reproducir (EXIT para salir): ")

    if texto.strip().lower() == "exit":
        break

    reproducir_texto(texto)
```

```python
from TTS.api import TTS
import numpy as np
import sounddevice as sd
import librosa
import torch
import re

# ============================
# CONFIGURACIÓN OPTIMIZADA
# ============================
CONFIG = {
    "speed": 1.02,
    "pitch": 0.3,
    "volume": 1.12,
    "sample_rate_out": 48000,
    "silence_trim": True,
    "normalize": True,
    "limit": True
}

device = "cuda" if torch.cuda.is_available() else "cpu"
tts = TTS("tts_models/es/css10/vits").to(device)

# ============================
# PROCESAMIENTO DE TEXTO
# ============================
def preprocess_text(texto):
    texto = texto.strip()
    texto = re.sub(r"\s+", " ", texto)
    texto = texto.replace(",", ", ")
    texto = texto.replace(".", ". ")
    return texto

# ============================
# RECORTAR SILENCIOS
# ============================
def trim_silence(audio, threshold=0.015):
    idx = np.where(np.abs(audio) > threshold)[0]
    if len(idx) == 0:
        return audio
    return audio[idx[0]:idx[-1]]

# ============================
# NORMALIZADOR DE picos
# ============================
def normalize_audio(audio):
    peak = np.max(np.abs(audio))
    if peak > 0:
        return audio / peak
    return audio

# ============================
# LIMITADOR ANTI-CLIPPING
# ============================
def limiter(audio, limit=0.97):
    audio = np.clip(audio, -limit, limit)
    return audio

# ============================
# FUNCIÓN PRINCIPAL
# ============================
def reproducir_texto(texto):
    texto = preprocess_text(texto)
    print("\nTexto:", texto)

    # 1. Generar audio
    audio = tts.tts(
        texto,
        speed=CONFIG["speed"],
        pitch=CONFIG["pitch"]
    )
    audio = np.array(audio, dtype=np.float32)

    # 2. Volumen
    audio *= CONFIG["volume"]

    # 3. Recortar silencios
    if CONFIG["silence_trim"]:
        audio = trim_silence(audio)

    # 4. Normalización
    if CONFIG["normalize"]:
        audio = normalize_audio(audio)

    # 5. Anti-clipping
    if CONFIG["limit"]:
        audio = limiter(audio)

    # 6. Upsample a 48 kHz
    audio = librosa.resample(audio, orig_sr=22050, target_sr=CONFIG["sample_rate_out"])

    # 7. Reproducir
    sd.play(audio, CONFIG["sample_rate_out"])
    sd.wait()

    print("Audio reproducido correctamente.")

# ============================
# LOOP INTERACTIVO
# ============================
while True:
    texto = input("\nEscribe el texto (EXIT para salir): ")

    if texto.lower().strip() == "exit":
        break

    reproducir_texto(texto)
```
